```python
from pathlib import Path
import json, sys
sys.dont_write_bytecode = True
import experiment as ex

RECOMPUTE = False  # True = rerun; False = load results/report.json when present
PEAK_TFLOPS = {mode: None for mode in ("FP64", "FP32", "TF32", "FP16")}
PEAK_SOURCE = ""  # Official specification URL for the actual GPU; dense peaks only
settings = ex.Settings(peaks_tflops=PEAK_TFLOPS, peak_source=PEAK_SOURCE)
result_file = Path("results/report.json")
report = ex.run(settings) if RECOMPUTE or not result_file.exists() else json.loads(result_file.read_text())
ex.apply_peaks(report, PEAK_TFLOPS, PEAK_SOURCE)
rendered = ex.pages(report)
ex.export_html(report, rendered=rendered)
ex.show(rendered, 1)
```
